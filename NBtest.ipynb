## **Quelques équations** 

L'équation de conservation de la masse s'écrit :  

$$ \dfrac{\partial \rho}{\partial t} + \vec{\nabla} \left( \rho \vec{v} \right) = 0 $$

L'équation d'onde s'écrit :

$$ \dfrac{\partial ^2 \vec{E}}{\partial t^2} = \dfrac{1}{c^2}\vec{\nabla}^2 \vec{E}$$

L'equation de Dirac s'écrit :

$$ (i\hbar\gamma^{\mu}\partial_{\mu} - mc)\psi = 0$$

L'équation de décroissance radioactive s'écrit :

$$ N(t) = N_0e^{-\lambda t}$$
